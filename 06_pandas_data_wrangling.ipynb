```python
import pandas as pd

data = {
    "name": ["Amina", "Brian", "Chao", "Dina", "Eli"],
    "score": [88, 72, 95, 67, 81],
    "hours": [10, 7, 12, 5, 9],
}

df = pd.DataFrame(data)
df
```

```python
# Filter and derive columns
filtered = df[df["score"] >= 80].copy()
filtered["efficiency"] = filtered["score"] / filtered["hours"]
filtered.sort_values("efficiency", ascending=False)
```

```python
# Grouping example
bins = pd.cut(df["hours"], bins=[0, 6, 9, 12], labels=["low", "mid", "high"])
df.groupby(bins)["score"].mean()
```

```python
import pandas as pd
import numpy as np

# First, CREATE the DataFrame
data = {
    'score': [95, 82, 67, 88, 91, 73, 85, 79, 94, 62],
    'hours': [8, 6, 5, 9, 10, 4, 7, 6, 12, 3]
}
df = pd.DataFrame(data)
print("Original DataFrame created:")
print(df)
print("\n" + "="*50 + "\n")

# Now you can do all the operations
# 1. Add pass/fail column
df["result"] = df["score"].apply(lambda x: "pass" if x >= 75 else "fail")
print("DataFrame with pass/fail column:")
print(df)
print("\n" + "="*50 + "\n")

# 2. Grouping example
bins = pd.cut(df["hours"], bins=[0, 6, 9, 12], labels=["low", "mid", "high"])
grouped_result = df.groupby(bins)["score"].mean()
print("Mean scores by hours category:")
print(grouped_result)
print("\n" + "="*50 + "\n")

# 3. Filter and derive columns
filtered = df[df["score"] >= 80].copy()
filtered["efficiency"] = filtered["score"] / filtered["hours"]
filtered_sorted = filtered.sort_values("efficiency", ascending=False)
print("Filtered (score >= 80) and sorted by efficiency:")
print(filtered_sorted)
```
