## Learning goals
- Create line, bar, and histogram charts
- Use visualization for storytelling

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sns.set_theme(style="whitegrid")

sales = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "revenue": [1200, 1350, 1280, 1500, 1650, 1720],
})

plt.figure(figsize=(8, 4))
plt.plot(sales["month"], sales["revenue"], marker="o")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
students = pd.DataFrame({
    "group": ["A", "A", "B", "B", "C", "C"],
    "score": [70, 75, 80, 85, 90, 95]
})

plt.figure(figsize=(6, 4))
sns.boxplot(data=students, x="group", y="score")
plt.title("Score Distribution by Group")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create the students dataframe
students = pd.DataFrame({
    "group": ["A", "A", "B", "B", "C", "C"],
    "score": [70, 75, 80, 85, 90, 95]
})

# Practice task 1: Create a histogram
plt.figure(figsize=(8, 4))
plt.hist(students["score"], bins=5, edgecolor='black', alpha=0.7)
plt.title("Distribution of Scores (Histogram)")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
print("This histogram shows how many students scored in each score range. Most scores are distributed across the range.")

# Add pass/fail column to students data
students["result"] = students["score"].apply(lambda x: "pass" if x >= 75 else "fail")
print("\nStudents data with pass/fail column:")
print(students)
```
